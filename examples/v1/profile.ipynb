# Profile

```python
profiles_path = "../../datasets/v1/profiles/dataset1/001.h5"
mean_profile_path = "../../datasets/v1/mean_profiles/dataset1/001.h5"
out_path = None
```

```python
from heavyedge import ProfileData
import matplotlib.pyplot as plt

with ProfileData(profiles_path) as data:
    Ys, Ls, _ = data[:]
with ProfileData(mean_profile_path) as data:
    (mean_Y,), (mean_L,), _ = data[:]

for Y, L in zip(Ys, Ls):
    plt.plot(Y[:L], color="gray", alpha=0.5)
plt.plot(mean_Y[:mean_L])

if out_path is not None:
    plt.savefig(out_path)
else:
    plt.show()
```
